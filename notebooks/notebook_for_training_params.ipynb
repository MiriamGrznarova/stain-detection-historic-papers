```python
!pip install ultralytics torch torchvision matplotlib seaborn pandas

import os
import torch
import matplotlib.pyplot as plt
import pandas as pd
from ultralytics import YOLO

device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'I am using this device:: {device}')

model = YOLO('yolov8m.pt')  # loading model (yolov8n/ yolov8s/ yolov8m)

data_path = "/kaggle/input/custom_model/data.yaml"

results = model.train(
    data=data_path,
    epochs=100,        
    imgsz=1280,        
    batch=8,           
    device=device,     
    save=True,         
    optimizer='SGD',   
    lr0=0.01,          
    lrf=0.0001,        
    momentum=0.937,    
    weight_decay=0.0005, 
    
    # Augmentation
    mosaic=1.0,        # Mosaic augmentation (helps with small objects)
    mixup=0.2,         # Mixup for better generalization
    fliplr=0.5,        # Horizontal rotation
    flipud=0.1,        # vertical rotation
    scale=0.5          # random scaling
)
```

```python
#validation
model = YOLO("/kaggle/working/runs/detect/train/weights/best.pt")
metrics = model.val(data=data_path)
print(metrics)
```

```python
# test data
metrics_test = model.val(data=data_path, split="test")
print(metrics_test)
```
