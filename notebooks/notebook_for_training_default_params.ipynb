```python
!pip install ultralytics torch torchvision matplotlib seaborn pandas
import os
import torch
import matplotlib.pyplot as plt
import pandas as pd
from ultralytics import YOLO

device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'I am using this device: {device}')
```

```python
model = YOLO("yolov8n.pt")  # loading model (yolov8n/ yolov8s/ yolov8m)
data_path = "/kaggle/input/custom_dataset/data.yaml"
results = model.train(
    data=data_path,
    epochs=100,        
    imgsz=1280,        
    batch=8,           # small memory
    device=device,     
    save=True
)
```

```python
# validation
model = YOLO("/kaggle/working/runs/detect/train/weights/best.pt")
metrics = model.val(data=data_path) 
print(metrics)
```

```python
# test data
metrics_test = model.val(data=data_path, split="test")
print(metrics_test)
```
